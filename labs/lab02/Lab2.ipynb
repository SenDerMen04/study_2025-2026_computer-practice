```julia
# Задание 3.14

using Random

x = rand(0:999, 250)
y = rand(0:999, 250)

filtered_y = y[y .> 600]
println("Элементы y > 600: ", filtered_y)

corresponding_x = x[findall(y .> 600)]
println("Соответ. x: ", corresponding_x)

sum_exp = sum(exp.(-x[2:end] + x[1:end-1]) .+ 10)
println("Сумма: ", sum_exp)

unique_x = unique(x)
println("Уник. элем. x: ", unique_x)
```
